```python
import os
import shutil
import subprocess
from google.colab import drive

# 1. Montar Google Drive
drive.mount('/content/drive')

# 2. Definir la ruta de la carpeta del Diplomado en tu Drive
# Ajusta la ruta si tu carpeta está dentro de otra subcarpeta
diplomado_path = '/content/drive/MyDrive/Diplomado/ModuloI/Codigos/Formato'

if not os.path.exists(diplomado_path):
    print(f"⚠️ No se encontró la ruta: {diplomado_path}")
    print("Verifica el nombre exacto de tu carpeta en Google Drive.")
else:
    print(f"📁 Procesando archivos en: {diplomado_path}\n")

    total_ipynb = 0
    total_py = 0

    # Recorrer todos los directorios y subdirectorios
    for root, dirs, files in os.walk(diplomado_path):
        for file in files:
            file_path = os.path.join(root, file)

            # Caso 1: Archivos Jupyter Notebook (.ipynb -> .md)
            if file.endswith('.ipynb') and not file.startswith('.'):
                base_name = os.path.splitext(file)[0]
                output_md = os.path.join(root, f"{base_name}.md")

                # Ejecutar nbconvert para generar Markdown limpio
                cmd = f'jupyter nbconvert --to markdown "{file_path}" --output-dir="{root}"'
                result = subprocess.run(cmd, shell=True, capture_output=True, text=True)

                if result.returncode == 0:
                    print(f"✅ [.ipynb -> .md] {file} -> {base_name}.md")
                    total_ipynb += 1
                else:
                    print(f"❌ Error al convertir {file}: {result.stderr}")

            # Caso 2: Scripts de Python (.py -> .txt)
            elif file.endswith('.py') and not file.endswith('.py.txt'):
                output_txt = f"{file_path}.txt"
                shutil.copyfile(file_path, output_txt)
                print(f"✅ [.py -> .txt]    {file} -> {os.path.basename(output_txt)}")
                total_py += 1

    print("\n" + "="*40)
    print(f"🎉 Proceso terminado:")
    print(f"   • Notebooks (.ipynb) convertidos a Markdown: {total_ipynb}")
    print(f"   • Scripts (.py) copiados a .txt: {total_py}")
    print("="*40)
```
